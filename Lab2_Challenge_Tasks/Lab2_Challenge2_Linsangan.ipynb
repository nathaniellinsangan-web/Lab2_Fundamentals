```python
import random
import string

SURNAME = "LINSANGAN"    
SEED_NUM = 6        
SIGNAL_LENGTH = 15   

# ==========================================
# 1. GENERATE SIGNAL
# ==========================================
# Seed based on surname and SEED_NUM for reproducible data generation
seed_value = sum(ord(char) for char in SURNAME) + SEED_NUM
random.seed(seed_value)

# Pool of characters: uppercase, lowercase, digits, spaces, and punctuation
char_pool = string.ascii_letters + string.digits + "   " + string.punctuation
generated_signal = "".join(random.choice(char_pool) for _ in range(SIGNAL_LENGTH))

# ==========================================
# 2. PROCESS AND NORMALIZE SIGNAL
# ==========================================
# Strip leading/trailing whitespaces and convert to uppercase for normalization
processed_signal = generated_signal.strip().upper()

# ==========================================
# 3. CHARACTER BY CHARACTER ANALYSIS
# ==========================================
execution_log = []
char_analysis = []

letters_count = 0
digits_count = 0
spaces_count = 0
special_count = 0

for idx, char in enumerate(processed_signal, start=1):
    if char.isalpha():
        char_type = "Letter"
        letters_count += 1
    elif char.isdigit():
        char_type = "Digit"
        digits_count += 1
    elif char.isspace():
        char_type = "Space"
        spaces_count += 1
    else:
        char_type = "Special Character"
        special_count += 1

    char_analysis.append((char, char_type))
    execution_log.append(f"Char {idx}: '{char}' -> {char_type}")

# ==========================================
# 4. SIGNAL CLASSIFICATION
# ==========================================
# Classify based on dominant character types present
total_chars = len(processed_signal)

if letters_count == total_chars:
    signal_classification = "Pure Text Signal"
elif digits_count == total_chars:
    signal_classification = "Pure Numeric Signal"
elif letters_count > 0 and digits_count > 0 and special_count == 0:
    signal_classification = "Alphanumeric Signal"
elif special_count > 0:
    signal_classification = "Complex / Noise-Corrupted Signal"
else:
    signal_classification = "Mixed Text Signal"

# ==========================================
# 5. ASSESSMENT DATA OUTPUT
# ==========================================
print("==========================================")
print("            ASSESSMENT DATA               ")
print("==========================================")
print(f"Generated Signal    : '{generated_signal}'")
print(f"Processed Signal    : '{processed_signal}'")
print(f"Signal Classification: {signal_classification}")

print("\nCharacter Analysis:")
for char, char_type in char_analysis:
    print(f"  '{char}' -> {char_type}")

print("\nExecution Log:")
for log in execution_log:
    print(f"  {log}")

print("\nFinal Output:")
print("  --- DIAGNOSTIC REPORT ---")
print(f"  Processed Length : {total_chars} characters")
print(f"  Letter Count     : {letters_count}")
print(f"  Digit Count      : {digits_count}")
print(f"  Space Count      : {spaces_count}")
print(f"  Special Char Count: {special_count}")
print(f"  Overall Status   : {signal_classification}")
print("==========================================")
```
