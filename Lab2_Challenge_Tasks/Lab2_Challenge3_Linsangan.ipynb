```python
import random
import string

# ==========================================
# CONFIGURATION / INPUT DATA
# ==========================================
SURNAME = "LINSANGAN"     # Replace with your surname
SEED_NUM = 6         # Replace with your assigned SEED_NUM

# ==========================================
# 1. GENERATE ACCESS CREDENTIAL & ATTEMPT LIMIT
# ==========================================
# Seed base using surname and SEED_NUM for reproducible results
seed_value = sum(ord(char) for char in SURNAME) + SEED_NUM
random.seed(seed_value)

# Generate a password based on surname and random alphanumeric suffix
pass_length = 6
char_pool = string.ascii_letters + string.digits
generated_password = f"{SURNAME[:3].upper()}{''.join(random.choice(char_pool) for _ in range(pass_length))}"

# Dynamically set attempt limit (e.g., between 3 and 5 attempts)
attempt_limit = random.randint(3, 5)

# Simulated login attempts (a mix of incorrect attempts and the correct password)
simulated_inputs = ["wrongpass1", "123456", generated_password, "another_pass"]

# ==========================================
# 2. AUTHENTICATION PROCESS & EXECUTION
# ==========================================
attempts_made = []
execution_log = []

access_result = "Access Denied"
final_system_state = "LOCKED OUT"

for idx, user_input in enumerate(simulated_inputs, start=1):
    if len(attempts_made) >= attempt_limit:
        execution_log.append(f"Attempt {idx} ignored: Maximum attempt limit ({attempt_limit}) reached.")
        break
    
    attempts_made.append(user_input)
    
    if user_input == generated_password:
        access_result = "Access Granted"
        final_system_state = "UNLOCKED"
        execution_log.append(f"Attempt {len(attempts_made)}: '{user_input}' -> SUCCESS. System Unlocked.")
        break
    else:
        execution_log.append(f"Attempt {len(attempts_made)}: '{user_input}' -> FAILED. Incorrect Password.")

if final_system_state == "LOCKED OUT":
    execution_log.append(f"Security Alert: Maximum allowed attempts ({attempt_limit}) reached. Authentication process terminated.")

# ==========================================
# 3. ASSESSMENT DATA OUTPUT
# ==========================================
print("==========================================")
print("            ASSESSMENT DATA               ")
print("==========================================")
print(f"Generated Password: {generated_password}")
print(f"Attempt Limit     : {attempt_limit}")
print(f"Attempts Made     : {len(attempts_made)}")
print(f"Access Result     : {access_result}")
print(f"Final System State: {final_system_state}")

print("\nExecution Log:")
for log in execution_log:
    print(f"  {log}")

print("\nFinal Output:")
print("  --- AUTHENTICATION SUMMARY ---")
print(f"  System Status      : {final_system_state}")
print(f"  Authentication     : {access_result}")
print(f"  Total Attempts Used: {len(attempts_made)} / {attempt_limit}")
print(f"  Attempt Details    : {attempts_made}")
print("==========================================")
```
